```python
# Flattening the DataFrame
def flatten(df):
    # Iterate through columns
    flat_cols = []
    for col_name in df.columns:
        # If the column is of struct type, flatten it
        if isinstance(df.schema[col_name].dataType, StructType):
            for field in df.schema[col_name].dataType.fields:
                flat_cols.append(col(col_name + '.' + field.name).alias(col_name + '_' + field.name))
        else:
            flat_cols.append(col(col_name))
   
    return df.select(flat_cols)
```

```python


# Apply the flatten function
flat_df = flatten(df)
flat_df.show(truncate=False)
```

```python

from pyspark.sql.functions import explode
# Explode the array
df_exploded = df.withColumn("better_call_saul_appearance", explode("better_call_saul_appearance"))

# Flatten after exploding
df_flat_exploded = flatten(df_exploded)
df_flat_exploded.show()
```
